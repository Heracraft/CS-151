```python
def has22(nums):
    for index, num in enumerate(nums):
        if (num==2 and nums[index+1]==2):
            return True
    return False

print(has22([1,6,2,2]))
print(has22([1,2,6,2,5,2,6]))
print(has22([1,2,6]))
print(has22([8,2, "2","2",7]))
```

```python
def sum13(nums):
    sum = 0
    if len(nums) == 0:
        return sum
    for index, num in enumerate(nums):
        # index-1 ,to get the previous element, always works.
        # except for the first element where index is 0
        if num == 13 or nums[index-1] == 13:
            # 0-1 = -1, giving you the value last element (nums[-1]) instead. So in our algorithm here, 
            # if the last element is 13 (edge case)
            continue  # The first item won't be added
        sum += num
    return sum


sum13([1, 2, 2, 1, 13])  # The funny edge case I was referring to
```

```python
def removeDuplicates(someString):
    finalChars=[]
    encounteredChars=[]
    for char in someString:
        if (char not in encounteredChars or char == " "):
            finalChars.append(char)
            encounteredChars.append(char)

    return str().join(finalChars)

print(removeDuplicates("soo manny duupplicatess"))
```

```python
from abc import ABC, abstractmethod

class Undead(ABC):
    
    @abstractmethod
    def speak(self):
        pass

class Vampire(Undead):
    pass
    
    def speak(self):
        print("I come for your blood")

class Ghost(Undead):
    def speak(self):
        print("Boo!")

class Zombie():
 
    def mumbles(self):
        print("BRaaaaAiNS!")

class UndeadCollection():
    def __init__(self):
        self.list = []

    def append(self,elem):
        if isinstance(elem, Undead):
            self.list.append(elem)
        else:
            print("error: can only add undead")

    def pop(self):
        return self.list.pop()

    def getUndead(self):
        return self.list[:]

d = Vampire()
c = Ghost()
c.speak()
d.speak()
monsters = UndeadCollection()
monsters.append(Vampire())
monsters.append(Ghost())
monsters.append(Zombie())
for m in monsters.getUndead():
    m.speak()
```

```python
class Pet:
    scientificClass = 'Mammalia'
    scientificOrder = 'Carnivora'
    legs = 4

    def __init__(self, name):
        self.name = name


class Dog(Pet):
    scientificFamily = 'Canidae'
    scientificGenus = 'Canis'

    def __init__(self, name, birthYear, breed="mutt"):
        super().__init__(name)

        self.breed = breed
        self.birthYear = birthYear
        self.tricks = []

    def addTrick(self, trickCommand, probability):
        self.tricks.append([trickCommand, probability])

    def getName(self):
        return self.name

    def setName(self, newName):
        self.name = newName

    def getAge(self):
        age = 2020 - self.birthYear
        return age

    def numberOfTricks(self):
        return len(self.tricks)


class Cat(Pet):
    scientificFamily = 'Felidae'
    scientificGenus = 'Felis'

    def __init__(self, name, weight, birthYear):
        super().__init__(name)

        self.weight = weight
        self.birthYear = birthYear

    def getWeight(self):
        return self.weight

    def setWeight(self, weight):
        self.weight = weight
```

```python
def letter_frequency(text):
    frequency={}
    for char in text:
        frequency[char]=text.count(char)
    return frequency
print(letter_frequency("Hello how are you?")) #{'H': 1, 'e': 2, 'l': 2, 'o': 3, ' ': 3, 'h': 1, 'w': 1, 'a': 1, 'r': 1, 'y': 1, 'u': 1, '?': 1}
```

```python
def getMode(items):
    highestFreqItem=items[1]
    highestFreq=0
    for item in items:
        if items.count(item)>highestFreq:
            highestFreqItem=item
            highestFreq=items.count(item)

    return highestFreqItem

print(getMode([2,5,6,4,7,4,4,2,4,6,8,9,1,2,7])) # 4
```

```python
def sum_num(num):
    if num == 0:
        return num + 0

    return num + sum_num(num-1)


sum_num(5)

# What is the base case?
# What is the recursive case?
# What is the recursive call?

# TestCode:
# sum_num(3)
# sum_num(10)

# Output:
# 3+2+1+0 = 6
```
