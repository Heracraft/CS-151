```python
def evenOdd(n):
	currentNum=0
	while currentNum <= n:
		print(f"{currentNum} is {"even" if currentNum%2==0 else "odd"}")
		currentNum+=1

evenOdd(4)
```

```python
def isBoth(items1, items2):
    return isinstance(items1, list) and isinstance(items2, list)

print(isBoth([2,3], "3"))
print(isBoth([2,3], [23,455,6]))
```

```python
def has22(nums):
    for index, num in enumerate(nums):
        if (num==2 and nums[index+1]==2):
            return True
    return False

print(has22([1,6,2,2]))
print(has22([1,2,6,2,5,2,6]))
print(has22([1,2,6]))
print(has22([8,2, "2","2",7]))
```
